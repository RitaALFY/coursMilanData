```python
import re
import string

def compter_mots(texte):
    return re.sub(f"[{[x for x in string.punctuation  if x not in "-'"]}]", " ", texte).split()
```

```python
# fonction qui renvoie le plus long mot dans le texte
def plus_long_mot(texte):
    tableau_mots = compter_mots(texte)
    long_mot = tableau_mots[0]

    for word in tableau_mots[1:]:
        if len(word) > len(long_mot):
            long_mot = word
    return long_mot


def plus_long_mot_avec_max(texte):
    tableau_mots = compter_mots(texte)
    mot = max(tableau_mots, key=len)
    return mot


def les_plus_longs_mots(texte):
    tableau_mots = compter_mots(texte)
    longueur = len(max(tableau_mots, key=len))
    resultat_mots = [mot for mot in tableau_mots if len(mot) == longueur]
    return resultat_mots

def les_plus_longs_mots_v2(texte):
    liste_mots = compter_mots(texte)
    long_mot = liste_mots[0]
    les_longs_mots = [long_mot]
    for mot in liste_mots[1:]:
        if len(mot) > len(long_mot):
            long_mot = mot
            les_longs_mots = [mot]
        elif len(mot) == len(long_mot):
            les_longs_mots.append(mot)
    return les_longs_mots

def longestWords(str:str):
    if not str : return -1
    listWords = compter_mots(str)
    
    listWordsTmp=[]
    wordLongest = listWords[0]

    for word in listWords[1:]:
        if len(word) >= len(wordLongest):
            wordLongest=word
            listWordsTmp.append(wordLongest)
            if len(listWordsTmp[0]) < len(wordLongest):
                listWordsTmp.pop(0)
    return list(set(listWordsTmp))
["aa", "bb", "ccc"]
["bb", "ccc"]

les_plus_longs_mots_v2("bonjour tout bontour le monde")
```

```python
# fonction qui renvoie la liste des nombres premiers allant de 2 a n

def liste_nombres_premiers(num):
    liste = []
    if num < 2:
        return -1
    for i in range(2, num+1):
        if est_premier_v3(i):
            liste.append(i)
    return liste

def liste_nombres_premiers_v2(num):
    return [element for element in range(2,num+1) if est_premier_v3(element)]

def liste_nombres_premiers_v3(num):
    # 6 * k +-1
    if num < 2:
        return []
    if num == 2:
        return [2]
    if num in [3, 4]:
        return [2, 3]
    i = 1
    liste = [2, 3]
    while i*6-1 <= num:
        if est_premier_v3(i*6-1):
            liste.append(i*6-1)
        if i*6+1 > num:
            break
        if est_premier_v3(i*6+1):
            liste.append(i*6+1)
        i += 1
    return liste

liste_nombres_premiers_v3(55)
```

```python
# fonction suite fibonacci
def suite_fibonacci(num):
    if type(num) is not int or (num <= 0):
        return -1
    if num == 1:
        return [0]
    elif num == 2:
        return [0, 1]
    else:
        liste = [0, 1]
        for _ in range(2, num):
            liste.append(liste[-1] + liste[-2])
        return liste


def suite_fibonacci_v2(num, index=2, liste=[0, 1]):
    if index >= num:
        return liste
    liste.append(liste[-1] + liste[-2])
    return suite_fibonacci_v2(num, index+1, liste)

suite_fibonacci_v2(10)
```

```python
# fonction qui fait un chiffrement de cesar
import string
import unicodedata
def chiffrement_cesar(texte, decalage):
    if (not texte) or type(decalage) is not int or decalage <=0:
        return -1
    alphabet_minuscule = list(string.ascii_lowercase)
    alphabet_majuscule = list(string.ascii_uppercase)

    liste_origine = list(texte)
    liste_finale = []

    for i in range(len(liste_origine)):
        if not liste_origine[i].isalpha():
            liste_finale.append(liste_origine[i])
        elif liste_origine[i].islower():
            position = alphabet_minuscule.index(liste_origine[i])
            position_decalage = (position + decalage) % 26
            new_lettre = alphabet_minuscule[position_decalage]
            liste_finale.append(new_lettre)
        else:
            position = alphabet_majuscule.index(liste_origine[i])
            position_decalage = (position + decalage) % 26
            new_lettre = alphabet_majuscule[position_decalage]
            liste_finale.append(new_lettre)
    return "".join(liste_finale)

def chiffrement_cesar_opti(texte, decalage):
    if (not texte) or type(decalage) is not int:
        return -1
    alphabet_minuscule = list(string.ascii_lowercase)
    alphabet_majuscule = list(string.ascii_uppercase)

    texte_chiffre = ""
    for caractere in texte:

        if not caractere.isalpha():
            texte_chiffre += caractere
        else: 
            if caractere.islower():
                alphabet = alphabet_minuscule
            else:
                alphabet = alphabet_majuscule
            position = alphabet.index(caractere)
            position_decalage = (alphabet.index(caractere) + decalage) % 26
            new_lettre = alphabet[position_decalage]
            texte_chiffre += new_lettre
    return texte_chiffre

def chiffrement_cesar_v2(texte, decalage):
    result = ""

    for char in texte:
        if 'a' <= char <= 'z':
            new_char = chr((ord(char) - ord('a') + decalage) % 26 + ord('a'))
            result += new_char
        elif 'A' <= char <= 'Z':
            new_char = chr((ord(char) - ord('A') + decalage) % 26 + ord('A'))
            result += new_char
        else:
            result += char
    return result

def chiffrement_cesar_v3(texte, decalage):
    result = ""

    for char in texte:
        if char.isalpha() and char.isascii():
            if char.islower():
                new_char = chr((ord(char) - ord('a') + decalage) % 26 + ord('a'))
                result += new_char
            else:
                new_char = chr((ord(char) - ord('A') + decalage) % 26 + ord('A'))
                result += new_char
        else:
            result += char
    return result

def chiffrement_cesar_v4(texte, decalage):
    result = ""

    for char in texte:
        if unicodedata.category(char) == "Ll":
            new_char = chr((ord(char) - ord('a') + decalage) % 26 + ord('a'))
            result += new_char
        elif unicodedata.category(char) == "Lu":
            new_char = chr((ord(char) - ord('A') + decalage) % 26 + ord('A'))
            result += new_char
        else:
            result += char
    return result

def dechiffrement_cesar(texte, decalage):
    result = ""

    for char in texte:
        if unicodedata.category(char) == "Ll":
            new_char = chr((ord(char) - ord('a') - decalage) % 26 + ord('a'))
            result += new_char
        elif unicodedata.category(char) == "Lu":
            new_char = chr((ord(char) - ord('A') - decalage) % 26 + ord('A'))
            result += new_char
        else:
            result += char
    return result

def chiffrer_ou_dechiffrer_cesar(texte:str, decalage:int, chiffrer:bool=True):
    """
    fonction qui va retourner le texte chiffré ou déchiffré

    exemple d'appel de fonction:
        chiffrer_ou_dechiffrer_cesar("Bcd.y za", 1, chiffrer=True)
    """
    if chiffrer:
        return chiffrement_cesar_v4(texte, decalage)
    else:
        return dechiffrement_cesar(texte, decalage)

# dechiffrement_cesar(chiffrement_cesar_v2("Abc.x yz", 1), 1)
chiffrer_ou_dechiffrer_cesar("Bcd.y za", 1, chiffrer=True)
    
```

```python
# fonction qui fusionne deux listes triees

import numpy as np
def fusion_listes_triees(l1, l2):
    nouvelle_liste = np.array(l1)
    indice = 0
    for element in l2:
        while indice < len(nouvelle_liste) and element > nouvelle_liste[indice]:
            indice += 1
        nouvelle_liste = np.insert(nouvelle_liste, indice, element)
    return nouvelle_liste

def fusion_listes_triees_v2(l1, l2):
    l3 = l1 + l2
    for i in range(len(l3)):
        for j in range(i+1, len(l3)):
            if l3[i] > l3[j]:
                l3[i], l3[j] = l3[j], l3[i]
    return l3

def fusion_listes_triees_v3(l1, l2):
    liste_fusionnee = []
    while l1 != [] or l2 != []:
        if l1 == []:
            liste_fusionnee.extend(l2)
            break
        elif l2 == []:
            liste_fusionnee.extend(l1)
            break
        else:
            if l1[0] <= l2[0]:
                liste_fusionnee.append(l1.pop(0))
            else:
                liste_fusionnee.append(l2.pop(0))
    return liste_fusionnee


fusion_listes_triees_v2([1,2,3,8,9], [2,2,4,5,6,7])
```

```python
# fonction qui renvoie un dictionnaire de lettres 
# dans l'ordre de la plus frequente a la moins frequente

def ordre_lettres_par_frequence(texte):
    if not texte:
        return -1
    dictionnaire = {}
    for lettre in texte:
        if dictionnaire.get(lettre): #if lettre in dictionnaire:
            dictionnaire[lettre] += 1
        else:
            dictionnaire[lettre] = 1
    return dict(sorted(dictionnaire.items(), key=lambda item: item[1], reverse=True))

import string
def ordre_lettres_par_frequence_v2(texte):
    if not texte:
        return -1
    dictionnaire = {}
    alphabet_minuscule = string.ascii_lowercase

    for element in alphabet_minuscule:
        compteur = texte.count(element)
        if compteur != 0:
            dictionnaire[element] = compteur
    return dict(sorted(dictionnaire.items(), key=lambda item: item[1], reverse=True))

from operator import itemgetter
def ordre_lettres_par_frequence_v3(texte):
    if not texte:
        return -1
    dictionnaire = {}
    for lettre in texte:
        if dictionnaire.get(lettre): #if lettre in dictionnaire:
            dictionnaire[lettre] += 1
        else:
            dictionnaire[lettre] = 1
    return dict(sorted(dictionnaire.items(), key=itemgetter(1, 0), reverse=True))

ordre_lettres_par_frequence_v3("bbaaact")
```

```python
# fonction qui compresse les caracteres repetes
# aaccdrraa : a2c2dr2a2

def compression_texte(texte):
    if not texte:
        return -1
    resultat = ""
    count = 1
    for i in range(1, len(texte)):
        if texte[i] == texte[i-1]:
            count += 1
        else:
            if count > 1:
                resultat += texte[i-1] + str(count)
                count = 1
            else:
                resultat += texte[i-1]
                count = 1
    if count > 1:
        resultat += texte[-1] + str(count)
    else:
        resultat += texte[-1]
    return resultat

def compression_texte_v2(texte):
    compteur = 1
    resultat = ""
    for i in range(len(texte)):
        if i+1 < len(texte) and texte[i] == texte[i+1]:
            compteur += 1
        else:
            resultat += texte[i]
            if compteur > 1:
                resultat += str(compteur)
            compteur = 1
    return resultat

from itertools import groupby

def compresser_chaine_v2(chaine: str) -> str:
    return ''.join(
        char + (str(len(list(groupe))) if len(list(groupe)) > 1 else '')
        for char, groupe in groupby(chaine)
    )

compression_texte_v2("aabbaaccdaa")
```

```python
#fonction qui calcule le plus grand conteneur
# [1,5,4,8,9,6,1]
# hauteur du plus petit mur multiplie par la distance entre les deux dans la liste

from functools import cache, lru_cache
import time

def conteneur_max(height):
    if type(height) is not list:
        return -1
    max_actuel = (0,0,0)

    for i in range(len(height)-1):
        for j in range(i+1, len(height)):
            plus_petit = min(height[i], height[j])
            surface = (i, j, (plus_petit * (j-i)))
            if surface[2] > max_actuel[2]:
                max_actuel = surface
            
    return max_actuel

@cache
def calcul(hg, hd, largeur):
    time.sleep(0.001)
    return (min(hg, hd)) * largeur

def conteneur_max_v2(height):
    gauche = 0
    droite = len(height) - 1
    maximum = 0
    indice1 = 0
    indice2 = 0
    while gauche < droite:
        conteneur = calcul(height[gauche], height[droite], droite - gauche)
        if conteneur > maximum:
            maximum = conteneur
            indice1 = gauche
            indice2 = droite
        if height[gauche] < height[droite]:
            gauche += 1
        else:
            droite -= 1
    return maximum, indice1, indice2
conteneur_max_v2([1,5,4,8,9,6,1])
```

```python
#fonction d'expressions regulières.
#les elements '.' et '*'
# vous recevez deux string, le premier a que des lettres minuscules
# le deuxieme a des lettres minuscules et peux contenir des '.' et '*'
# on suppose que chaque '*' a avant lui un caractere

def expression_a_tester(s, p):
    stock = {}
    def compare_caractere(i, j):

        # On retourne la valeur stockée afin d'éviter les boucles si i et j sont stocké
        if (i, j) in stock:
            return stock[(i, j)]
        
        # on arrive à la fin du masque mais pas à la fin de la chaine
        if j == len(p):
            return i == len(s)
                
        # on compare si i est pas la fin de s et que i = j ou que i = .
        match = False
        if i < len(s) and (s[i] == p[j] or p[j] == '.'):
            match = True

        # on regarde si le caractere apres j est toujours dans p et que ce caratere = *
        resultat = False
        if j + 1 < len(p) and p[j + 1] == '*':
            # on utilise les 2 options (garder ou laisser)
            # option 1 : on ignore *
            ignorer = compare_caractere(i, j + 2)

            # option 2 : on utilise le caractère
            garder = False
            if match and compare_caractere(i + 1, j):
                garder = True
            
            # Si on a soit option 1 soit option 2
            if ignorer or garder:
                resultat = True
        else:
            # Si cas sans étoile, on avance
            if match and compare_caractere(i + 1, j + 1):
                resultat = True

        stock[(i, j)] = resultat
        
        return resultat
    
    return compare_caractere(0, 0)
print(expression_a_tester("a", "b*a"))
```
