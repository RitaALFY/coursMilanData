## Difference de vitesse

```python
import time
liste = list(range(10_000_000))
a = np.arange(10_000_000)

#liste
t0 = time.perf_counter()
resultat_liste = [x*2 for x in liste]
t1 = time.perf_counter()

#array
t2 = time.perf_counter()
resultat_array = a * 2
t3 = time.perf_counter()

print(f"Liste Python : {(t1 - t0)*1000:.2f} ms")
print(f"NumPy        : {(t3 - t2)*1000:.2f} ms")
print(f"Accélération : {(t1 - t0)/(t3 - t2):.1f}x")
```

```python
# On vous donne les notes de 25 eleves sur 4 évaluations (notées /20).

rng = np.random.default_rng(75)
eval1 = rng.normal(11, 4, 25).round(1).clip(0, 20)
eval2 = rng.normal(12, 3, 25).round(1).clip(0, 20)
eval3 = rng.normal(13, 5, 25).round(1).clip(0, 20)
eval4 = rng.normal(10, 4, 25).round(1).clip(0, 20)

print(eval1)
print(eval2)
print(eval3)
print(eval4)


# 1. Créer une matrice 25x4 à partir des 4 vecteurs d'évaluations
notes = np.column_stack([eval1, eval2, eval3, eval4])
notes = np.matrix_transpose([eval1, eval2, eval3, eval4])
notes = np.array([eval1, eval2, eval3, eval4]).T
notes = np.transpose([eval1, eval2, eval3, eval4])
notes = np.stack([eval1, eval2, eval3, eval4], axis=1)
notes = np.hstack((eval1.reshape(-1,1), eval2.reshape(-1,1), eval3.reshape(-1,1), eval4.reshape(-1,1)))
print(notes)

# 2. Afficher la forme de la matrice
print(notes.shape)
# 3. Calculer et afficher : Moyenne générale, Écart-type général, Note minimale et maximale
print(f"Q3 : moyenne {np.mean(notes):.2f}, ecart-type {np.std(notes):.2f}, min/max {np.min(notes)}/{np.max(notes)}")
# 4. Pour chaque évaluation, calculer : moyenne, médiane, min, max
print(f"Q4 : moyennes= {notes.mean(axis=0)}, médianes={np.median(notes, axis=0)}")
print(f"Q4 : min={np.min(notes, axis=0)}, max={np.max(notes, axis=0)}")
# 5. Calculer la moyenne de chaque eleve
moyenne_eleve = np.mean(notes, axis=1)
print(f"Q5 : moyenne eleve={moyenne_eleve}")
# 6. Combien ont une moyenne >= 10 ?
moyenne_dessus_10 = moyenne_eleve >= 10
print(f"Q6 : {sum(moyenne_dessus_10)}")
print(f"Q6 : {sum(moyenne_eleve >= 10)}")
# 7. Quels sont les indices des 3 meilleurs apprenants ?
print(f"Q7 : {np.argsort(moyenne_eleve)[-3:]}")
print(f"Q7 : {np.argpartition(moyenne_eleve,-3)[-3:]}")
# 8. Quels apprenants ont eu >= 15 à l'évaluation 3 ?
print(f"Q8 : {np.where(notes[:,2] >= 15, notes[:,2], None)}")
print(f"Q8 : {np.where(notes[:,2])}")
print(f"Q8 : {np.argwhere(notes[:,2] >= 15)}")
# 9. Créer un masque pour les apprenants ayant TOUTES les notes >= 8
masque = np.all(notes >= 8, axis=1)
print(f"Q9 : {masque}")
print(f"{np.where(masque)}")
# 10. Bonus de 2 points sur les notes inférieures a 8 uniquement
notes[notes < 8] += 2
# notes = np.where(notes >= 8, notes, notes+2)
# notes = np.where(notes < 8, notes+2, notes)
print(f"Q10 : {notes}")
# 11. Bonus de 5 points sur l'évaluation 1 (plafonné à 20)
# notes[:,0] = np.clip((notes[:,0] + 5), 0, 20)
notes[:,0] += 5
# notes[:,0] = np.clip(notes[:,0], None, 20)
# notes[:,0] = np.clip(notes[:,0], max=20)
notes[:,0] = np.minimum(notes[:,0],20)

print(f"Q11 : {notes}")
# 12. Appliquer la normalisation Min-Max sur les notes de l'évaluation 1
#v1
notes_min = notes[:,0].min()
notes_max = notes[:,0].max()
eval1_norm = []
for une_note in notes[:,0]:
    eval1_norm.append((une_note - notes_min) / (notes_max - notes_min))
eval1_norm = np.array(eval1_norm)

#v2
notes_min = notes[:,0].min()
notes_max = notes[:,0].max()
eval1_norm = ((notes[:,0] - notes_min) / (notes_max - notes_min))

#v3
eval1_norm = ((notes[:,0] - notes[:,0].min()) / (notes[:,0].max() - notes[:,0].min()))

print(f"Q12 : {eval1_norm}")

# 13. Standardiser les notes de l'évaluation 2
eval2_standardized = (notes[:,1]-np.mean(notes[:,1])) / np.std(notes[:,1])
print(f"Q13 : {eval2_standardized}")
print(f"Q13 : mean={eval2_standardized.mean():.10f}   std={eval2_standardized.std()}")
# 14. Afficher un array de la meilleure note dans chaque matiere entre les 10 meilleurs eleves de la classe
print(f"Q14 : les moyennes : {notes.mean(axis=1)}")
print(f"Q14 : moyennes croissantes : {np.sort(notes.mean(axis=1))}")
top10_idx = np.argsort(notes.mean(axis=1))[-10:]
top10_notes = notes[top10_idx]
meilleures_notes = top10_notes.max(axis=0)
print(f"Q14 : {meilleures_notes}")

top10_idx = np.argpartition(notes.mean(axis=1), -10)[-10:]
top10_notes = notes[top10_idx]
meilleures_notes = top10_notes.max(axis=0)
print(f"Q14 : {meilleures_notes}")

meilleures_notes = notes[np.argpartition(notes.mean(axis=1), -10)[-10:]].max(axis=0)
print(f"Q14 : {meilleures_notes}")
```

### Exercice 4

```python
data = {
    "eleve": ["Léa", "Hugo", "Sarah", "Maxime", "Inès", "Paul"],
    "python":    ["18", 14, np.nan, 11, 16, 9],
    "sql":       [15, np.nan, 17, 8, np.nan, 12],
    "stats":     [12, 16.0, 14, 10, 15, np.nan],
    "present":   [True, True, True, True, True, False]
}
df = pd.DataFrame(data)
# 1. Faites en sorte que la matière "python" soit de type entier ou flotant
print(df.dtypes)
df["python"] = df["python"].astype(float)
df["python"] = pd.to_numeric(df["python"], errors="coerce")
print(f"Q1 : {df.dtypes}")
print(f"Q1 : {df}")

# 2. Diagnostiquer les valeurs manquantes
print(f"Q2 : \n{df.isna().sum()}")
# 3. Remplir les notes manquantes avec la moyenne de leur matière
# 1e maniere
df.fillna({
    "python": df["python"].mean(),
    "sql": df["sql"].mean(),
    "stats": df["stats"].mean()
    }, inplace=True)
# 2e maniere
df.fillna(df.mean(numeric_only=True), inplace=True)
#3e maniere
df["python"] = df["python"].fillna(df["python"].mean().round(2))
df["sql"] = df["sql"].fillna(df["sql"].mean().round(2))
df["stats"] = df["stats"].fillna(df["stats"].mean().round(2))
print(f"Q3 : \n{df}")
# 4. Créer une colonne "moyenne" (moyenne des 3 matières)
df["moyenne"] = df[["python", "sql", "stats"]].mean(axis=1, skipna=True).round(1)
print(f"Q4 : \n{df}")
# 5. Créer une colonne "statut" : "Validé" si moyenne >= 10, "Rattrapage" si entre 8 et 10, "Échec" sinon
#1e maniere
def statut_moyenne(moyenne):
    if moyenne >= 10:
        return "Validé"
    elif moyenne >= 8:
        return "Rattrapage"
    else:
        return "Echec"
df["statut"] = df["moyenne"].apply(statut_moyenne)
print(f"Q5 : \n{df}")
#2e maniere
borne = [float("-inf"), 8, 10, float("inf")]
label = ["Echec", "Rattrapage", "Validé"]
df["statut"] = pd.cut(df["moyenne"], bins=borne, labels=label, right=False)
print(f"Q5 : \n{df}")
#3e maniere
conditions = [(df["moyenne"] >= 10), (df["moyenne"]>= 8)]
value = ["Validé", "Rattrapage"]
df["statut"] = np.select(conditions, value, default="Echec")
df["statut"] = np.where(df["moyenne"] >= 10, "Validé", np.where(df["moyenne"] >= 8, "Rattrapage", "Echec"))
print(f"Q5 : \n{df}")
# 4e maniere
def statut_moyenne(moyenne):
    if moyenne >= 10:
        return "Validé"
    elif moyenne >= 8:
        return "Rattrapage"
    else:
        return "Echec"
df["statut"] = df["moyenne"].map(statut_moyenne)
print(f"Q5 : \n{df}")
#5e maniere
df.loc[df["moyenne"] >= 10, "statut"] = "Validé"
df.loc[df["moyenne"] < 8, "statut"] = "Echec"
df.loc[(df["moyenne"] >= 8) & (df["moyenne"] < 10), "statut"] = "Rattrapage"
print(f"Q5 : \n{df}")

# 6. Supprimer la colonne "present"
df.drop(["present"], axis=1, inplace=True)
print(f"Q6 : \n{df}")
# 7. Compter combien d'eleves sont "Validé"
print(f"Q7 : {(df["statut"] == "Validé").sum()}")
# 8. Quelle est la moyenne générale de la promo ?
print(f"Q8 : {df["moyenne"].mean()}")
```

```python
# Création du dataset
rng = np.random.default_rng(101)
sites = ["Roanne"] * 10 + ["St-Étienne"] * 10 + ["Lyon"] * 10
niveaux = rng.choice(["Débutant", "Intermédiaire", "Avancé"], 30, p=[0.5, 0.3, 0.2])
note = rng.normal(14, 4, 30).round(1).clip(0, 20)
heures_etude = rng.integers(5, 30, 30)
abandon = rng.choice([0, 1], 30, p=[0.8, 0.2])
df = pd.DataFrame({
    "site": sites,
    "niveau": niveaux,
    "note": note,
    "heures_etude": heures_etude,
    "abandon": abandon
})
print(df)
# 1. Note moyenne par site
print(f"Q1 : \n{df.groupby("site")["note"].mean()}")
# 2. Statistiques par site ("mean", "std", "min", "max", "count")
print(f"Q2 : \n{df.groupby("site")["note"].agg(["mean", "std", "min", "max", "count"]).round(2)}")
# 3. Note moyenne par site ET par niveau
print(f"Q3 : \n{df.groupby(["site", "niveau"])["note"].agg(["mean"])}")
print(f"Q3 : \n{df.groupby(["site", "niveau"])["note"].mean()}")
print(f"Q3 : \n{df.groupby(["site", "niveau"]).agg(note_moyenne=pd.NamedAgg("note", "mean"))}")
# 4. Nombre d'apprenants par site
print(f"Q4 : \n{df.groupby(["site"])["niveau"].agg(["count"])}")
print(f"Q4 : \n{df["site"].value_counts()}")
print(f"Q4 : \n{df.groupby("site").size()}")
print(f"Q4 : \n{df.groupby("site")["site"].count()}")
# df["apprenant"] = 1
# print(f"Q4 : \n{df.groupby("site")["apprenant"].sum()}")
nombre_apprenants_par_site = df.groupby("site").size().reset_index(name="nombre_apprenants")
print(f"Q4 : \n{nombre_apprenants_par_site}")
# 5. Taux d'abandon par site
print(f"Q5 : \n{df.groupby("site")[["abandon"]].mean()*100}")
# 6. Heures d'étude moyennes par niveau
print(f"Q6 : \n{df.groupby("niveau")["heures_etude"].mean().round(2)}")
# 7. Classement des sites avec la meilleure note moyenne
print(f"Q7 : {df.groupby("site")["note"].mean().sort_values(ascending=False)}")
```
