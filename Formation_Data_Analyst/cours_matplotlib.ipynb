```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

# plt.plot(x, y)
# plt.plot(x, y, color="red")
# plt.plot(x, y, color="red", marker="*")
# plt.plot(x, y, color="red", marker="*", markersize=20)
# plt.plot(x, y, color="red", marker="*", markersize=20, markerfacecolor="#151515")
# plt.plot(x, y, color="red", marker="*", markersize=20, mfc="#151515")
# plt.plot(x, y, color="red", marker="*", 
#          markersize=20, mfc="#151515", linestyle="dashed")
plt.plot(x, y, color="red", marker="*", 
         markersize=20, mfc="#151515", markeredgecolor="#06FA3B", linestyle="dashed", linewidth=4)
plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([10, 30, 31, 25])

plt.figure(figsize=(18, 5), dpi=200)
# plt.plot(x, y)
# plt.plot(x, y, color="red")
# plt.plot(x, y, color="red", marker="*")
# plt.plot(x, y, color="red", marker="*", markersize=20)
# plt.plot(x, y, color="red", marker="*", markersize=20, markerfacecolor="#151515")
# plt.plot(x, y, color="red", marker="*", markersize=20, mfc="#151515")
# plt.plot(x, y, color="red", marker="*", 
#          markersize=20, mfc="#151515", linestyle="dashed")
plt.plot(x, y1, color="red", marker="*", 
         markersize=20, mfc="#151515", markeredgecolor="#06FA3B", linestyle="dashed", linewidth=4)
plt.plot(x, y2, color="blue", marker="*", 
         markersize=20, mfc="#151515", mec="#06FA3B", linestyle="dashed", linewidth=4)
plt.title("C'est un graphe")
plt.xlabel("label x")
plt.ylabel("label y")
plt.xticks(x, ["année 1", "année 2", "année 3", "année 4"])
# plt.xticks(x)
# plt.xticks(np.arange(2023, 2026.25, 0.25))
plt.grid()
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.plot(x, y1)
plt.title("Premier graphique")

plt.subplot(1, 2, 2)
plt.plot(x, y2)
plt.title("Deuxieme graphique")
```

```python
fig, axs = plt.subplots(2,2, figsize=(10,8))

axs[0, 0].plot(x, y1)
axs[0, 0].grid()
axs[0, 1].scatter(x, y1)
axs[1, 0].bar(x, y1)
axs[1, 0].set_xticks(x)
fig.suptitle("mes graphes")
fig.tight_layout()
# fig.savefig("fig.png")
```

```python
fig, ax = plt.subplots(figsize=(8,5))
ax.plot(x, y1, 'o-', color="darkgreen", linewidth=2)
ax.grid(alpha=0.5, linestyle='--')
ax.set_xlim(2023, 2026)
ax.axhline(26)
plt.show()
```
